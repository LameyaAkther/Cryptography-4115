## 2. Greatest common divisor and Euclid's algorithm\(\gcd(a,b)\)\[\]\((a,b)\) \((b,a\bmod b)\)\(x,y\)\[\]

```python
def gcd(a, b):
    a, b = abs(a), abs(b)
    while b:
        a, b = b, a % b
    return a

def extended_gcd(a, b):
    # Return (d, x, y) such that d = gcd(a,b) = a*x + b*y.
    if b == 0:
        return (abs(a), 1 if a >= 0 else -1, 0)
    d, x1, y1 = extended_gcd(b, a % b)
    x = y1
    y = x1 - (a // b) * y1
    return d, x, y

print("gcd(24,16) =", gcd(24,16))
print("extended_gcd(27,15) =", extended_gcd(27,15))
d, x, y = extended_gcd(27, 15)
assert d == gcd(27,15) and 27*x + 15*y == d
```

## 3. Least common multiple\(a,b\)\[\]\[\]

```python
def lcm(a, b):
    if a <= 0 or b <= 0:
        raise ValueError("a and b must be positive")
    return (a // gcd(a, b)) * b

assert lcm(9, 3) == 9
assert lcm(17, 37) == 629
print(lcm(12, 18))
```

## 4. Linear Diophantine equations\[\]\(\gcd(a,b)\mid c\)\(ax_0+by_0=d\) \(d=\gcd(a,b)\) \(c/d\)\[\]

```python
def diophantine(a, b, c):
    d, x0, y0 = extended_gcd(a, b)
    if c % d != 0:
        raise ValueError("No integer solution exists")
    scale = c // d
    x, y = x0 * scale, y0 * scale
    assert a*x + b*y == c
    return x, y

print("34x + 51y = 170:", diophantine(34, 51, 170))
print("91x + 117y = 65:", diophantine(91, 117, 65))
```

## 5. Modular inverse and modular division\(a\) \(n\) \(\gcd(a,n)=1\) \(a^{-1}\)\[\]\(ax\equiv b\pmod n\) \(x\equiv ba^{-1}\pmod n\)\(a\) \(ax+ny=1\)

```python
def mod_inverse(a, n):
    if n <= 1:
        raise ValueError("Modulus must be greater than 1")
    d, x, _ = extended_gcd(a, n)
    if d != 1:
        raise ValueError("Inverse does not exist because gcd(a,n) != 1")
    return x % n

def modular_divide(b, a, n):
    return (b * mod_inverse(a, n)) % n

assert (5 * mod_inverse(5, 54)) % 54 == 1
print("9 / 5 mod 3:", modular_divide(9, 5, 3))
print("223 / 67 mod 54:", modular_divide(223, 67, 54))
```

## 6. Fast modular exponentiation\(b^e\bmod m\) \(b^e\) \(O(\log e)\)\(m\)

```python
def fast_modular_exponentiation(b, e, m):
    if e < 0 or m <= 0:
        raise ValueError("Require e >= 0 and m > 0")
    result = 1 % m
    base = b % m
    exponent = e
    while exponent > 0:
        if exponent % 2 == 1:
            result = (result * base) % m
        base = (base * base) % m
        exponent //= 2
    return result

assert fast_modular_exponentiation(4, 5, 3) == 1
assert fast_modular_exponentiation(7, 13, 11) == pow(7,13,11)
print(fast_modular_exponentiation(7, 128, 11))
```

## 7. Chinese Remainder Theorem (two moduli)\(n_1,n_2\)\[\]\(n_1n_2\)\(n_1x+n_2y=1\)\[\]

```python
def chinese_remainder_theorem(n1, r1, n2, r2):
    if n1 <= 0 or n2 <= 0 or gcd(n1, n2) != 1:
        raise ValueError("Moduli must be positive and coprime")
    _, x, y = extended_gcd(n1, n2)  # n1*x + n2*y = 1
    result = (r1*n2*y + r2*n1*x) % (n1*n2)
    assert result % n1 == r1 % n1
    assert result % n2 == r2 % n2
    return result

assert chinese_remainder_theorem(13, 7, 6, 5) == 59
assert chinese_remainder_theorem(28, 20, 19, 3) == 440
print(chinese_remainder_theorem(13, 7, 6, 5))
```

## 8. Fermat's Little Theorem and Euler's theorem\(p\) \(p\nmid a\)\[\]\(\phi(n)\) \(1\) \(n\) \(n\) \(p,q\)\[\]\(\gcd(a,n)=1\)\[\]

```python
def euler_totient(n):
    if n < 1:
        raise ValueError("n must be positive")
    count = 0
    for k in range(1, n+1):
        if gcd(k, n) == 1:
            count += 1
    return count

p = 13
a = 2
assert fast_modular_exponentiation(a, p-1, p) == 1
print("phi(12) =", euler_totient(12))
print("2^(12) mod 13 =", fast_modular_exponentiation(2, 12, 13))
```

## 9. Prime numbers and primality testing\(\sqrt n\)

```python
def is_prime_trial(n):
    if n < 2:
        return False
    if n % 2 == 0:
        return n == 2
    d = 3
    while d*d <= n:
        if n % d == 0:
            return False
        d += 2
    return True

print([(n, is_prime_trial(n)) for n in range(1, 20)])
```

Additional Day 2 slide topics### Infinitude and distribution of primes\(4k+1\) \(4k+3\) \(\gcd(a,b)=1\) \(ak+b\) \(\pi(x)\) \(x/\ln x\)### Prime-number conjectures\(n^2+1\)### Sieve of Eratosthenes\(n\) \(\sqrt n\)### Fermat pseudoprimes and Carmichael numbers\(a^{n-1}\equiv1\pmod n\) \(\gcd(a,n)=1\) \(a^n\equiv a\pmod n\) \(a\)\(p\) \(p-1\mid n-1\)### Primitive roots, Lucas tests and Mersenne primes\(n-1\) \(2^p-1\)

```python
def sieve_of_eratosthenes(limit):
    if limit < 2:
        return []
    is_prime = [True] * (limit + 1)
    is_prime[0] = is_prime[1] = False
    p = 2
    while p*p <= limit:
        if is_prime[p]:
            for multiple in range(p*p, limit+1, p):
                is_prime[multiple] = False
        p += 1
    return [n for n, flag in enumerate(is_prime) if flag]

print("Primes up to 50:", sieve_of_eratosthenes(50))
assert sieve_of_eratosthenes(20) == [2,3,5,7,11,13,17,19]
```

```python
def fermat_probable_prime(n, base):
    # Fermat test is only a compositeness test/probable-prime test,
    # not a proof that n is prime.
    if n < 2 or gcd(base, n) != 1:
        return False
    return pow(base, n-1, n) == 1

print("341, base 2:", fermat_probable_prime(341, 2))
print("341, base 3:", fermat_probable_prime(341, 3))
print("91, base 3:", fermat_probable_prime(91, 3))
print("91, base 2:", fermat_probable_prime(91, 2))
print("561, base 2:", fermat_probable_prime(561, 2))
print("561, base 5:", fermat_probable_prime(561, 5))
```

# Number Theory – Day 1 Assignment

$GCD$ $(a, b)$$(d, x, y)$ $d = GCD(a, b)$ $(x, y)$ $d = ax + by$ $a \geq b \geq 0$

```python
def extended_gcd(a, b):
    if b == 0:
        return (a, 1, 0)

    d, x1, y1 = extended_gcd(b, a % b)

    x = y1
    y = x1 - (a // b) * y1

    return (d, x, y)
```

```python
print(extended_gcd(8, 4))
print(extended_gcd(27, 15))
```

$a > 0$ $b > 0$ $c$ $x$ $y$ $ax + by = c$ $a, b ,c$

```python
def diophantine(a, b, c):
    gcd, x0, y0 = extended_gcd(a, b)

    assert c % gcd == 0

    k = c // gcd
    x = x0 * k
    y = y0 * k

    assert a * x + b * y == c

    return (x, y)
```

```python
print(diophantine(34, 51, 170))
print(diophantine(91, 117, 65))
```

$b$ $a$ $n$$a$ $b$ $n$ $GCD(a, n) = 1$ $n > 1$ $x$$0 \leq x \leq n-1$$b  /  a = x \pmod{n}$ $b = ax \pmod{n}$$a \pmod{n}$

```python
def modInverse(a, n):
    gcd, x, y = extended_gcd(a, n)

    assert gcd == 1

    return x % n


def divide(b, a, n):
    assert n > 1 and a > 0

    aInv = modInverse(a, n)

    assert (a * aInv) % n == 1

    return (aInv * b) % n
```

```python
print(divide(9, 5, 3))
print(divide(223, 67, 54))
```

$b^e \bmod{m}$ $2\log_2{e}$

```python
def FastModularExponentiation(b, e, m):
    result = 1
    b = b % m

    while e > 0:
        if e % 2 == 1:
            result = (result * b) % m

        b = (b * b) % m
        e = e // 2

    return result
```

```python
print(FastModularExponentiation(4, 5, 3))
```

$ n_1, r_1, n_2, r_2 $ $n_1$ $n_2$ $( n_1 > n_2 )$ $0 \leq r_1 < n_1$ $0 \leq r_2  < n_2$ $r$ $0 \leq r < n_1n_2$ $r \equiv r_1 \bmod{n_1}$ $r \equiv r_2 \bmod{n_2}$

```python
def ChineseRemainderTheorem(n1, r1, n2, r2):
    d, x, y = extended_gcd(n1, n2)

    assert d == 1

    result = (r1 * n2 * y + r2 * n1 * x) % (n1 * n2)

    return result
```

```python
print(ChineseRemainderTheorem(13, 7, 6, 5)) # 59
print(ChineseRemainderTheorem(28, 20, 19, 3)) # 440
```
